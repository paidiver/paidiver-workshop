# Benthic Dataset

## Why this pipeline applies each step

1. **Open a small subset of Haig Fras images.** The JPEG images provide the visual record. The iFDO metadata links each image to its acquisition time and location. Fixed sampling gives you ten images to inspect while you tune the workflow.
2. **Convert to grayscale.** Compare seabed texture and brightness without the colour cast. Keep the originals for interpretations that depend on colour.
3. **Filter by acquisition time.** The example keeps images from a selected time interval. Inspect the timestamps in the metadata table before changing the limits. A narrower interval may leave few or no images.
4. **Blur gently, then sharpen.** Blur reduces fine fluctuations. Sharpening aims to make edges easier to see, but can amplify remaining noise or create halos. Compare the intermediate outputs to judge the trade-off.
5. **Adjust contrast.** Local contrast can reveal faint seabed texture. Check dark background areas for amplified noise.
6. **Crop the centre.** The example keeps a central region to study a smaller field of view. Check what is lost near each edge before applying the crop to scientific measurements.

Inspect the same image at successive steps. The metadata table after the run can show which files passed the time filter. Export after inspecting the result.

This tutorial features an image dataset from a 2012 survey in Haig Fras, UK.  

The dataset consists of raw seafloor visual images captured during a survey within the Greater Haig Fras Marine Conservation Zone (MCZ) in the central Celtic Sea. This was the first in a series of surveys conducted at this location. A camera system mounted on the autonomous underwater vehicle (AUV) Autosub6000 was deployed during the RRS Discovery cruise 377/8 (D377/8), capturing images along four 4.7 km transect lines.

## Configuration file

This is the configuration file used in this jupyter notebook


```yaml
general:
  input_path: "/groups/paidiver/images/benthic_haigfras"
  output_path: "output/benthic"
  metadata_path: "metadata/metadata_benthic_ifdo.json"
  metadata_type: "IFDO"
  image_open_args: "JPEG"
  n_jobs: 8
  sampling:
    - mode: "fixed"
      params:
        value: 10
steps:
  - colour:
      name: "colour_correction"
      mode: "grayscale"

  - sampling:
      name: "datetime"
      mode: "datetime"
      params:
        min: "2016-05-11 04:14:00"
        max: "2016-05-11 09:26:12"

  - colour:
      name: "colour_correction"
      mode: "gaussian_blur"
      params:
        sigma: 1.0

  - colour:
      name: "sharpen"
      mode: "sharpen"
      params:
        alpha: 1.5
        beta: 0.5

  - colour:
      name: "contrast"
      mode: "contrast"

  - convert:
      name: "crop_images"
      mode: "crop"
      params:
        size: 0.5
        size_type: "percent"
        mode: "center"
```
