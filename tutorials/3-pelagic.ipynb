# Pelagic Dataset

## Why this pipeline applies each step

1. **Open a small BMP subset.** The pelagic images are paired with CSV metadata. Fixed sampling keeps ten files so you can repeat the exercise quickly.
2. **Correct uneven illumination.** Rolling-ball correction estimates slowly varying background brightness. Look for a flatter background and check that faint organisms are still visible.
3. **Increase local contrast.** CLAHE makes faint structure easier to distinguish from its surroundings. The clip limit constrains the enhancement. Too much contrast can make background grain resemble features.
4. **Smooth gently.** A small Gaussian blur reduces some fine grain after contrast enhancement. Watch for disappearing thin structures. Compare the same image after each step before deciding whether to keep the blur.

Try changing a single parameter in a copy of the YAML, then create a fresh pipeline and rerun. Compare the same filename. Tutorial 4 introduces custom and library algorithms separately.

This tutorial features an image dataset collected in 2023, during the DY157 RSS Discovery Cruise.

The images were collected using a red camera frame deployed vertically via winches.

## Configuration file

This is the configuration file used in this jupyter notebook


```yaml
general:
  input_path: "/groups/paidiver/images/pelagic"
  output_path: "output/pelagic"
  metadata_path: "metadata/metadata_pelagic.csv"
  metadata_type: "CSV_FILE"
  image_open_args: "BMP"
  n_jobs: 1
  sampling:
    - mode: fixed
      params:
        value: 10
steps:
  - colour:
      name: even_lighting
      mode: illumination_correction
      params:
        method: rolling
        radius: 10
  - colour:
      name: local_contrast
      mode: contrast
      params:
        method: clahe
        clip_limit: 0.01
  - colour:
      name: gentle_blur
      mode: gaussian_blur
      params:
        sigma: 1.0
```
